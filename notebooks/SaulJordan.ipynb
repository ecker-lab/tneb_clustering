```python
import corc.utils
import numpy as np
import time
```

```python
# tmm = corc.utils.load_tmms("noisy_moons")[0]
tmm = corc.utils.load_tmms("densired16")[0]
```

```python
mixture_model = tmm.mixture_model
```

```python
gaussians = list()
for i in range(len(mixture_model.weights)):

    precision = np.linalg.inv(mixture_model.covs[i])
    gaussians.append(corc.SaulJordan.Gaussian(mixture_model.centers[i], precision))
```

```python
path_optimizer = corc.SaulJordan.PathOptimizer(
    mixture_model.centers[0],
    mixture_model.centers[1],
    gaussians
)
```

```python
start_time = time.time()
qts, zts = path_optimizer.iterate_and_solve_via_paper_approach()
print(f"Optimization of 50 iterations took {time.time()-start_time:.2f} seconds.")
```
