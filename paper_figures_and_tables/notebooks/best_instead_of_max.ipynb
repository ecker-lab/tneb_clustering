```python
import corc.graph_metrics.neb
import corc.utils
import corc.our_datasets
import os
import pickle
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import tqdm

cache_path="../../cache"
```

```python
filename = os.path.join(cache_path, "metrics", "figure2_main-full.pkl")
with open(filename, "rb") as f:
    data = pickle.load(f)
```

```python
datasets = corc.our_datasets.CORE_HD_DATASETS
dataset_name = datasets[6]
print(dataset_name)
X,y,tsne = corc.utils.load_dataset(dataset_name, cache_path=cache_path)
```

```python
filename = os.path.join(cache_path, f"{dataset_name}_TMM-NEB.pickle")
with open(filename, "rb") as f:
    tmms = pickle.load(f)
```

```python
avg_log_likelihoods = [-np.mean(tmm.mixture_model.score_samples(X)) for tmm in tmms]
aris = data[dataset_name]["TMM-NEB"]
np.vstack((avg_log_likelihoods,aris)).T
```

```python
np.corrcoef(avg_log_likelihoods, aris)[1,0]
```

```python
correlations = dict()
for dataset_name in tqdm.tqdm(datasets):
    X,y,tsne = corc.utils.load_dataset(dataset_name, cache_path=cache_path)
    filename = os.path.join(cache_path, f"{dataset_name}_TMM-NEB.pickle")
    with open(filename, "rb") as f:
        tmms = pickle.load(f)
    avg_log_likelihoods = [np.mean(tmm.mixture_model.score_samples(X)) for tmm in tmms]
    aris = data[dataset_name]["TMM-NEB"]
    correlations[dataset_name] = np.corrcoef(avg_log_likelihoods, aris)[1,0]
```
