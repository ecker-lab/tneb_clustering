```python
import corc.utils
# import corc
import matplotlib.pyplot as plt
import pickle
import corc.graph_metrics.uniforce
import corc.graph_metrics.neb
import numpy as np
import sklearn.metrics
import tqdm
import corc.our_datasets
import corc.create_datasets.complex_datasets
```

```python
X,y = corc.create_datasets.complex_datasets.load_densired(dim=384, path="../../datasets/densired-hd.npz")
tmm = corc.graph_metrics.neb.NEB(X, y, dataset_name="densired-hd", tmm_regularization=0.001, n_init=5, batch_size=30, optimization_iterations=100)
tmm.fit(X)
y_pred = tmm.predict_with_target(X, target_number_classes=6)
ari = sklearn.metrics.adjusted_rand_score(y, y_pred)
ari
```
